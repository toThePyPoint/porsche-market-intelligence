```python
with open("html-files/preview.html", "r", encoding="utf-8") as f:
    result = f.read()

html = BeautifulSoup(result, "html.parser")
page_items = html.find_all(class_=LAST_ITEMS_ID)
```

```python
search_results = html.find("div", {"data-testid": "search-results"})
articles = search_results.find_all("article", attrs={"data-id": True}, recursive=False)
```

```python
article = articles[1]
```

```python
title_element = article.find("h2").find("a")
```

```python
title = title_element.get_text(strip=True)
```
